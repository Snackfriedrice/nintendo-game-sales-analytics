```python
import pandas as pd

df = pd.read_csv("raw/vgsales.csv")
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=['Year'])

df['Publisher'] = df['Publisher'].fillna('Unknown')
```

```python
print(df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df['Year'] = df['Year'].astype(int)
```

```python
df['Calculated_Sales'] = (
    df['NA_Sales'] +
    df['EU_Sales'] +
    df['JP_Sales'] +
    df['Other_Sales']
)

df['Sales_Difference'] = abs(
    df['Global_Sales'] - df['Calculated_Sales']
)

print(df['Sales_Difference'].describe())
```

```python
print((df['Sales_Difference'] > 0.010001).sum())

print(
    df[df['Sales_Difference'] > 0.010001][
        ['Name', 'Global_Sales', 'Calculated_Sales', 'Sales_Difference']
    ]
)
```

```python
sales_cols = [
    'NA_Sales',
    'EU_Sales',
    'JP_Sales',
    'Other_Sales',
    'Global_Sales'
]

print((df[sales_cols] < 0).sum())

print(df['Year'].min())
print(df['Year'].max())
```

```python
df = df.drop(columns=['Calculated_Sales', 'Sales_Difference'])
```

```python
df.to_csv('cleaned/vgsales_cleaned.csv', index=False)
```
